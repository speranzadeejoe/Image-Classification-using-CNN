```python
import os
import shutil
import random

# Define dataset path
dataset_path = "/root/.cache/kagglehub/datasets/alxmamaev/flowers-recognition/versions/2/flowers"

# Define train and test paths
train_path = "/root/.cache/kagglehub/datasets/alxmamaev/flowers-recognition/versions/2/train"
test_path = "/root/.cache/kagglehub/datasets/alxmamaev/flowers-recognition/versions/2/test"

# Remove existing train/test folders to start fresh
shutil.rmtree(train_path, ignore_errors=True)
shutil.rmtree(test_path, ignore_errors=True)

# Create new directories
os.makedirs(train_path, exist_ok=True)
os.makedirs(test_path, exist_ok=True)

# Iterate through each class folder and split images
for class_folder in os.listdir(dataset_path):
    class_path = os.path.join(dataset_path, class_folder)

    if os.path.isdir(class_path):  # Ensure it's a folder
        images = os.listdir(class_path)
        random.shuffle(images)

        split_idx = int(0.7 * len(images))  # 70% for training

        # Define new class folder paths
        train_class_path = os.path.join(train_path, class_folder)
        test_class_path = os.path.join(test_path, class_folder)

        os.makedirs(train_class_path, exist_ok=True)
        os.makedirs(test_class_path, exist_ok=True)

        # Copy images to respective folders
        for img in images[:split_idx]:
            shutil.copy(os.path.join(class_path, img), os.path.join(train_class_path, img))

        for img in images[split_idx:]:
            shutil.copy(os.path.join(class_path, img), os.path.join(test_class_path, img))

print("Dataset split complete!")

# Verify number of images in train/test
print("Train dataset count:", sum(len(files) for _, _, files in os.walk(train_path)))
print("Test dataset count:", sum(len(files) for _, _, files in os.walk(test_path)))
```

```python

from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
# Construct an instance of the ImageDataGenerator
train_datagen = ImageDataGenerator(
    rescale = 1./255.,
    horizontal_flip = True,
    rotation_range = 20,
    width_shift_range = 0.2,
    height_shift_range = 0.2
)

test_datagen = ImageDataGenerator(
    rescale = 1./255
)
```

```python
from tensorflow.keras.applications import VGG16
from tensorflow.keras import layers, models
from tensorflow.keras.models import Model

# Load VGG16 base model
base_model = VGG16(include_top=False, input_shape=(224, 224, 3))

# Freeze base layers
for layer in base_model.layers:
    layer.trainable = False

# Add new classification layers
flatten_layer = layers.Flatten()(base_model.output)
fc_layer = layers.Dense(512, activation='relu')(flatten_layer)
output_layer = layers.Dense(5, activation='softmax')(fc_layer)

# Define final model
model = Model(inputs=base_model.input, outputs=output_layer)

training_iterator = train_datagen.flow_from_directory('/root/.cache/kagglehub/datasets/alxmamaev/flowers-recognition/versions/2/train', batch_size=64,
                                                      target_size=(224, 224))
# prepare iterators

testing_iterator = test_datagen.flow_from_directory('/root/.cache/kagglehub/datasets/alxmamaev/flowers-recognition/versions/2/test', batch_size=64,
                                                      target_size=(224, 224))

# Compile the model
model.compile(loss="categorical_crossentropy", optimizer="adam", metrics=["accuracy"])

# Fit model
history = model.fit(training_iterator, validation_data=testing_iterator, epochs=2)

# Display model summary
model.summary()
```

```python
import matplotlib.pyplot as plt

# Plot Loss vs Epochs
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('Model Loss')
plt.ylabel('Loss')
plt.xlabel('No of Epochs')
plt.legend(['Train', 'Test'], loc='upper left')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Plot Accuracy vs Epochs
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('Model Accuracy')
plt.ylabel('Accuracy')
plt.xlabel('No of Epochs')
plt.legend(['Train', 'Test'], loc='upper left')
plt.show()
```
